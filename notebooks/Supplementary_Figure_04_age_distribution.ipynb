```python
# Supplementary Figure 4 - Age distribution of IPD by serotype
# (previously held in a notebook named Figure_03)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
from scipy import stats

SOURCE = "data/Source_Data.xlsx"   # published with the paper; see data/README.md
OUT    = "results/figures/Supplementary_Figure_04.pdf"

df = pd.read_excel(SOURCE, sheet_name="Supplementary Figure 4")
s2    = df.loc[df["Group"] == "Serotype 2",      "Age (months)"].dropna()
other = df.loc[df["Group"] == "Other serotypes", "Age (months)"].dropna()

print(f"Serotype 2:      n={len(s2):5d}  median={s2.median():.1f}  "
      f"IQR={s2.quantile(.25):.1f}-{s2.quantile(.75):.1f}")
print(f"Other serotypes: n={len(other):5d}  median={other.median():.1f}  "
      f"IQR={other.quantile(.25):.1f}-{other.quantile(.75):.1f}")
u, p = stats.mannwhitneyu(s2, other, alternative="two-sided")
print(f"Wilcoxon rank-sum: U={u:.0f}, two-sided P={p:.3g}")
```

```python
plt.figure(figsize=(20, 12))
# seaborn >= 0.12 replaced shade= with fill=
sns.kdeplot(x=s2,    label="Serotype 2",      fill=True)
sns.kdeplot(x=other, label="Other Serotypes", fill=True)

plt.axvline(s2.median(),    color="blue",   linestyle="--", linewidth=1)
plt.axvline(other.median(), color="orange", linestyle="--", linewidth=1)

handles = [
    mlines.Line2D([], [], color="blue", linestyle="--", linewidth=1,
                  label=f"Median Serotype 2: {s2.median():.1f} months"),
    mlines.Line2D([], [], color="orange", linestyle="--", linewidth=1,
                  label=f"Median Other Serotypes: {other.median():.1f} months"),
]
plt.xlabel("Age (months)", fontsize=20); plt.ylabel("Density", fontsize=20)
plt.legend(handles=handles, loc="upper right", fontsize=20)
plt.xlim(0, 205); plt.xticks(fontsize=18); plt.yticks(fontsize=18)
plt.savefig(OUT, dpi=300, bbox_inches="tight")
plt.show()
```
