```python
# Figure 1 - Temporal trends in serotype 2 IPD, Bangladesh 1993-2022
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

SOURCE = "data/Source_Data.xlsx"   # published with the paper; see data/README.md
OUT    = "results/figures/Figure_01.pdf"

df = pd.read_excel(SOURCE, sheet_name="Figure 1")
df.head()
```

```python
fig, ax1 = plt.subplots(figsize=(20, 12))

ax1.bar(df["Year"], df["Serotype 2 detected by culture"],
        color="#67B2EC", label="Blood/CSF culture")
ax1.bar(df["Year"], df["Serotype 2 detected by culture-negative methods (ICT or PCR)"],
        bottom=df["Serotype 2 detected by culture"],
        color="#001d3d", label="Culture-negative (ICT or PCR)")

ax1.set_xlabel("Year", fontsize=26)
ax1.set_ylabel("Number of serotype 2 isolates", color="#001d3d", fontsize=26)
ax1.tick_params(axis="x", labelsize=22)
ax1.tick_params(axis="y", labelcolor="#001d3d", labelsize=22)
ax1.yaxis.set_major_locator(MaxNLocator(integer=True))

ax2 = ax1.twinx()
ax2.plot(df["Year"], df["Serotype 2 as % of meningeal IPD"],
         color="#F8D210", marker="o", linewidth=3, label="% of meningeal IPD")
ax2.plot(df["Year"], df["Serotype 2 as % of non-meningeal IPD"],
         color="grey", marker="o", linewidth=3, label="% of non-meningeal IPD")
ax2.set_ylabel("Serotype 2 as % of IPD", fontsize=26)
ax2.tick_params(axis="y", labelsize=22)

h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax1.legend(h1 + h2, l1 + l2, fontsize=18, loc="upper right")

fig.tight_layout()
plt.savefig(OUT, dpi=300, bbox_inches="tight")
plt.show()

print("Total serotype 2 cases:", int(df["Total serotype 2 cases"].sum()))
```
