```python
# Figure 2b - serotype distribution, IPD versus nasopharyngeal carriage cases
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

SOURCE = "data/Source_Data.xlsx"   # published with the paper; see data/README.md
SHEET  = "Figure 2b"
OUT    = "results/figures/Figure_02b.pdf"

df = pd.read_excel(SOURCE, sheet_name=SHEET)
df["Serotype"] = df["Serotype"].astype(str)
df.head()
```

```python
XCOL, YCOL = "Carriage_count", "Count_IPD"
COLOURS = {"Serotype 2": "red", "Covered by PCVs": "#F8D210", "Others": "blue"}

plt.figure(figsize=(25, 16))
for cat, colour in COLOURS.items():
    m = df["Category"] == cat
    plt.scatter(df.loc[m, XCOL], df.loc[m, YCOL],
                s=df.loc[m, "Bubble_size_total"] * 10,
                alpha=0.5, color=colour, label=cat)

for i in range(len(df)):
    off = (df["Bubble_size_total"].iloc[i] ** 0.5) * 0.05
    plt.text(df[XCOL].iloc[i] + off, df[YCOL].iloc[i] + off,
             df["Serotype"].iloc[i], fontsize=15, ha="center")

plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.xticks(fontsize=30); plt.yticks(fontsize=30)
plt.xlabel("Nasopharyngeal Carriage Cases", fontsize=30)
plt.ylabel("Invasive Pneumococcal Disease (IPD) Cases", fontsize=30)
plt.grid(True); plt.legend(fontsize=20, loc="lower right")
plt.savefig(OUT, dpi=300, bbox_inches="tight")
plt.show()

print(f"serotypes plotted: {len(df)}")
print(df[df["Serotype"] == "2"].to_string(index=False))
```
